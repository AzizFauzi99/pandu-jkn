# Query data NIK

Sumber: tabel mentah `bpjs_kepesertaan`, NIK ada di kolom `pstv01`. Tabel ini berisi jutaan baris, jadi selalu pakai `limit`.

```python
from utils.db import run_select
```

## Daftar NIK beserta status kepesertaan

```python
nik_data = run_select(
    """
    select k.pstv01 as nik,
           status.label_text as status,
           segmen.label_text as segmen,
           kelas.label_text as kelas_rawat
    from bpjs_kepesertaan k
    left join bpjs_metadata_lookup status
           on status.variable_name = 'PSTV17' and status.code_value = k.pstv17
    left join bpjs_metadata_lookup segmen
           on segmen.variable_name = 'PSTV08' and segmen.code_value = k.pstv08
    left join bpjs_metadata_lookup kelas
           on kelas.variable_name = 'PSTV07' and kelas.code_value = k.pstv07
    order by k.pstv01
    limit 100
    """,
    limit=100,
)
nik_data
```

## Detail satu NIK

```python
nik_cari = nik_data[0]["nik"]  # ganti dengan NIK yang dicari

detail_data = run_select(
    """
    select k.pstv01 as nik,
           k.pstv03 as tanggal_lahir,
           jenis_kelamin.label_text as jenis_kelamin,
           provinsi.label_text as provinsi,
           status.label_text as status,
           segmen.label_text as segmen,
           kelas.label_text as kelas_rawat
    from bpjs_kepesertaan k
    left join bpjs_metadata_lookup jenis_kelamin
           on jenis_kelamin.variable_name = 'PSTV05' and jenis_kelamin.code_value = k.pstv05
    left join bpjs_metadata_lookup provinsi
           on provinsi.variable_name = 'PSTV09' and provinsi.code_value = k.pstv09
    left join bpjs_metadata_lookup status
           on status.variable_name = 'PSTV17' and status.code_value = k.pstv17
    left join bpjs_metadata_lookup segmen
           on segmen.variable_name = 'PSTV08' and segmen.code_value = k.pstv08
    left join bpjs_metadata_lookup kelas
           on kelas.variable_name = 'PSTV07' and kelas.code_value = k.pstv07
    where k.pstv01 = %s
    """,
    (nik_cari,),
)
detail_data
```
